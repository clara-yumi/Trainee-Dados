```python
def verifica(list: list[int]):
    alertas: list[int] = []
    for i in range(len(list)):
        if list[i] > 100:
            alertas.append(list[i])
    if len(alertas) == 0:
        print('Sistema Seguro')
    else:
        print(f'Revisar Valores {alertas}')


valores_brutos = [10, 50, 150, 20, 300]
verifica(valores_brutos)
```
